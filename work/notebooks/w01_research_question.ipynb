```python
print("Total pages:", df["content_id"].nunique())

print("Total clients:", df["client_id"].nunique())

print("\nTrend direction:")
print(df["trend_direction"].value_counts())
```

```python
# Pages with at least 500 impressions
high_visibility = df[df["impressions_90d"] >= 500]

# Pages not updated for at least 180 days
old_pages = df[df["days_since_last_update"] >= 180]

# Pages that are both highly visible and not recently updated
stale_visible_pages = df[
    (df["impressions_90d"] >= 500) &
    (df["days_since_last_update"] >= 180)
]

print("Pages with 500+ impressions:", len(high_visibility))

print("Pages not updated in 180+ days:", len(old_pages))

print("Pages with 500+ impressions AND not updated in 180+ days:",
      len(stale_visible_pages))
```

Cost of a wrong recommendation:

False positive: A page that does not need attention is recommended, resulting in wasted review time and potentially unnecessary content updates.
False negative: A page that deserves attention is not recommended, potentially causing the team to miss an opportunity to investigate declining visibility or improve content performance.

The system will support human decision-making rather than automatically deciding which pages must be updated. Recommendations will be treated as opportunities for investigation, not guarantees of improved search performance.
